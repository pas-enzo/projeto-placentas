```python
import sys
import os
import torch
import ultralytics
import pandas as pd
import cv2
import numpy as np

# 1. Basic Version Check
print("--- BASIC VERSIONS ---")
print(f"Python version:      {sys.version.split()[0]}")
print(f"Ultralytics version: {ultralytics.__version__}")
print(f"PyTorch version:     {torch.__version__}")
print(f"Pandas version:      {pd.__version__}")
print(f"OpenCV version:      {cv2.__version__}")

# 2. GPU / CUDA Check
print("\n--- GPU / CUDA STATUS ---")
cuda_available = torch.cuda.is_available()
print(f"CUDA Available:      {cuda_available}")

if cuda_available:
    device_id = torch.cuda.current_device()
    props = torch.cuda.get_device_properties(device_id)
    print(f"GPU Name:            {torch.cuda.get_device_name(0)}")
    print(f"VRAM Total:          {props.total_memory / 1024**2:.0f} MB")
    print(f"CUDA Capability:     {props.major}.{props.minor}")
else:
    print("WARNING: GPU not detected by PyTorch. Training will be extremely slow!")

# 3. Path Verification (Ensuring you are on D:)
print("\n--- PATH VERIFICATION ---")
torch_path = torch.__file__
ultra_path = ultralytics.__file__

print(f"Torch Path:          {torch_path}")
print(f"Ultralytics Path:    {ultra_path}")

if "D:" in torch_path:
    print("SUCCESS: Environment is running from the D: drive.")
else:
    print("NOTE: Environment appears to be on another drive (check your VS Code kernel).")

# 4. Ultralytics Specific Check
print("\n--- ULTRALYTICS DIAGNOSTICS ---")
ultralytics.utils.checks.collect_system_info()
```
