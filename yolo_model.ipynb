```python
import yaml

data = {
    'path': r"D:/projeto_placentas_clayton/dataset_ready_for_yolo",
    'train': 'train/images',
    'val': 'valid/images',
    'names': {
        0: 'microcotiledone'
    }
}

# Save to file
yaml_path = r"D:\projeto_placentas_clayton\data.yaml"
with open(yaml_path, 'w') as f:
    yaml.dump(data, f, default_flow_style=False)

print(f"YAML file created at: {yaml_path}")
```

```python
from ultralytics import YOLO

model = YOLO('yolov8s-seg.pt')  

results = model.train(
    data=r"D:\projeto_placentas_clayton\data.yaml",
    epochs=100,
    imgsz=1024,
    batch=2,
    workers=2,
    patience=15,
    device=0,
    name='microcotiledone_run'
)
```
